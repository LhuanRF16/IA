```python
from sklearn import datasets
digits = datasets.load_digits()

from ucimlrepo import fetch_ucirepo

# fetch dataset
wine = fetch_ucirepo(id=109)

# data (as pandas dataframes)
wine_data = wine.data.features
wine_target = wine.data.targets
```

```python
from sklearn import tree
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

decision_tree = tree.DecisionTreeClassifier()
knn = KNeighborsClassifier()
naive_bayes = GaussianNB()

scaler = MinMaxScaler()
scaler.fit(wine_data)
wine_data_normalized = scaler.transform(wine_data)

wine_target = wine_target.to_numpy()
```

```python
from sklearn import model_selection
import numpy as np
kf = model_selection.StratifiedKFold(n_splits=3)

predicted_classes = dict()
predicted_classes['tree'] = np.zeros(wine_target.shape[0])
predicted_classes['knn'] = np.zeros(wine_target.shape[0])
predicted_classes['naive'] = np.zeros(wine_target.shape[0])

wine_data
```

```python
for train,test in kf.split(wine_data_normalized,wine_target):
    print(train)
    print(test)
```

```python
for train,test in kf.split(wine_data,wine_target):
    data_train,target_train = wine_data_normalized[train],wine_target[train]
    data_test,target_test = wine_data_normalized[test],wine_target[test]

    decision_tree = decision_tree.fit(data_train,target_train)
    decision_tree_predicted = decision_tree.predict(data_test)
    predicted_classes['tree'][test] = decision_tree_predicted

    knn = knn.fit(data_train,target_train)
    knn_predicted = knn.predict(data_test)
    predicted_classes['knn'][test] = knn_predicted

    naive_bayes = naive_bayes.fit(data_train,target_train)
    naive_predicted = naive_bayes.predict(data_test)
    predicted_classes['naive'][test] = naive_predicted
```

```python
from sklearn import metrics

for classifier in predicted_classes.keys():
    print("==========================================================================================================================")
    print("Resultados do classificador %s\n%s\n" %(classifier,metrics.classification_report(wine_target,predicted_classes[classifier])))
    print("Matriz de confusao: \n%s\n\n\n" %metrics.confusion_matrix(wine_target,predicted_classes[classifier]))
```
